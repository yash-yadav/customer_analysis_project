```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"F:\enviroment\python_sql_bi_project\customer_shopping_behavior.csv")
df.head()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
```

```python
df.rename(columns={"purchase_amount_(usd)" : "purchase_amount"}, inplace=True)
```

```python
df['review_rating'] = df['review_rating'].fillna(df.groupby('category')['review_rating'].transform('median'))
```

```python
df.isna().sum()
```

```python
# create a column age group

labels = ['young_adult', 'adult', 'middle_age', 'senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age_group', 'age']].head()
```

```python
# create column purchase_frequency_days

frequency_mapping = {
'Fortnightly': 14,
'Weekly': 7,
'Monthly': 30,
'Quarterly': 90,
'Bi-Weekly': 14,
'Annually': 365,
'Every 3 Months': 90

}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
